6. Explain the concept of generators in Python and how they are defined.

Ans- Generators in Python – Concept & Definition
A generator in Python is a special type of function used to generate a sequence of values lazily, meaning one at a time, as needed, instead of computing all results at once and storing them in memory.

Key Features of Generators:
Defined like a normal function, but uses the **yield** keyword instead of return.

Automatically implements the iterator protocol (__iter__() and __next__()).

More memory-efficient than lists for large data sequences.

 When to Use:
You don’t need all values at once.

You’re working with large data streams.

You want to reduce memory usage


```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
counter = count_up_to(5)

for num in counter:
    print(num)
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Ans-
1. map(function, iterable)
Purpose: Transforms each item in the iterable by applying a function.

Returns: A new iterable with transformed values.

2. filter(function, iterable)
Purpose: Filters elements by keeping those for which the function returns True.

Returns: A new iterable with only the "filtered" items.

3. reduce(function, iterable) (from functools)
Purpose: Applies a rolling computation to reduce the iterable to a single value.

Returns: A single value, not an iterable.



```python
from functools import reduce

nums = [1, 2, 3, 4]
print(list(map(lambda x: x*2, nums)))        # [2, 4, 6, 8]
print(list(filter(lambda x: x % 2 == 0, nums)))  # [2, 4]
print(reduce(lambda x, y: x + y, nums))      # 10
```

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

```python
def sum_even_numbers(numbers):
    """
    Returns the sum of all even numbers in the given list.
    """
    return sum(num for num in numbers if num % 2 == 0)

# Example usage
sample_list = [1, 2, 3, 4, 5, 6]
result = sum_even_numbers(sample_list)
print("Sum of even numbers:", result)
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    """
    Returns True if n is a prime number, else False.
    """
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Check and print all prime numbers from 1 to 200
prime_numbers = [i for i in range(1, 201) if is_prime(i)]
print("Prime numbers from 1 to 200:")
print(prime_numbers)
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
def fibonacci(n):
  """Generates the Fibonacci sequence up to n terms"""
  a,b=0,1
  for _ in range(n):
    yield a
    a, b = b,a+b

fib = fibonacci(10)
print("fibonacci sequence(10term):")
for num in fib:
  print(num,end='')
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(n):
    """
    Generator that yields powers of 2 from 0 to n (inclusive).
    """
    for i in range(n + 1):
        yield 2 ** i

# Example usage
print("Powers of 2 up to exponent 5:")
for power in powers_of_two(5):
    print(power, end=' ')
```

7. Implement a generator function that reads a file line by line and yields each line as a string.

```python


def read_file_lines(filepath):
    """
    Generator function that reads a file line by line.
    Yields each line as a string (stripped of newline).
    """
    with open(filepath, 'r') as file:
        for line in file:
            yield line.strip()

for line in read_file_lines('sample.txt'):
    print(line)
```

10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def remove_vowels(text):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda ch: ch not in vowels, text))

# Example usage
input_str = "Hello World"
output_str = remove_vowels(input_str)

print("Original String:", input_str)
print("Without Vowels:", output_str)
```
